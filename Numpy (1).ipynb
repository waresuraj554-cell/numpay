```python
# Element-wise operations
arr = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

# Addition
print(arr + arr2)

# Multiplication
print(arr * arr2)

# Universal functions (ufuncs)
print(np.sqrt(arr2))  # Square root of each element
```

```python
# Array with arange
arr = np.arange(100000, 1000000)
```

```python
# Reshape the array
reshaped_arr = arr.reshape(3000, 300)
```

```python
arr
```

```python
reshaped_arr
```

```python
from datetime import datetime
```

```python
datetime.now()
```

```python
start = datetime.now()
# Using Loop
flatten_using_loop_arr = []
for each in reshaped_arr:
    flatten_using_loop_arr.extend(each)
end = datetime.now()
print(f"time taken: {end-start}")
```

```python
flatten_using_loop_arr
```

```python
# Flatten the matrix using numpy
start = datetime.now()
flatten_using_numpy_arr = reshaped_arr.flatten()
end = datetime.now()
print(f"time taken: {end-start}")
```
