**Median Using Numpy**

- For meadian list must be sorted. Numpy median sorts the list automatically.
- If the dataset is in odd form it gets the middle value
- If the dataset is in even format it get two consecutive middle value sums it up and gives average as median of the dataset.

```python
arr = np.array([2, 16, 8, 10, 12])
# 2, 4, 8, 10, 12, 16
median_value = np.median(arr)
print(median_value)
```

**Get all the mode using**

```python
l = [1, 3, 3, 3, 2, 2, 2, 4, 4, 4, 5, 7]
```

```python
l
```

```python
d = {}
for each in l:
    if each in d:
        d[each] = d[each] + 1
    else:
        d[each] = 1
```

```python
d
```

```python
largest = 1
for k, v in d.items():
    if v > 1:
        largest = v
        
all_modes = []
for k, v in d.items():
    if v == largest:
        all_modes.append(k)
```

```python
largest
```

```python
all_modes
```

```python
print(f"All modes are: {sorted(all_modes)}, count: {largest}")
```

**Other numpy func**

```python
arr = np.array([1, 2, 3])
print(arr + 10)  # Output: [11 12 13]
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
```

```python
print(np.sum(arr))  # Output: 21
print(np.sum(arr, axis=0))  # Output: [5 7 9]
print(np.sum(arr, axis=1))  # Output: [ 6 15]
```
